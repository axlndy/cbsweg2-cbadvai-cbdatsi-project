## 1. Dataset Description
**Dataset Overview & Structure**  
A group of researchers gathered demographic information, socioeconomic indicators, behavioral habits, and learning perceptions from students and alumni of the University of Education – Vietnam National University, Hanoi. The survey was designed to investigate factors that may influence student learning outcomes. Data was collected between March and June* 2023 via a Google Form questionnaire, which used the snowball sampling method. The survey originally had 2,238 responses; however, "incomplete or inappropriate" submissions were dropped prior to publication. As a result, the dataset contains **2,170 validated observations (rows)** and **22 variables (columns)**. Each row represents a survey respondent, while columns correspond to specific questions.

*Note: The accompanying paper contains a minor inconsistency regarding the data collection period. The abstract states that responses were collected from March to June 2023, while the methodology section states that the survey was conducted from March to May 2023. Since no clarification is provided by the authors, this study only notes that data collection occurred during the March–June 2023 period.

**Implications of the Data Collection Method**  
While the dataset provides a robust sample size, several limitations and design choices impact result interpretation:

* **Sampling Bias:** The survey utilized voluntary participation and snowball sampling. Highly engaged students or specific social circles may be overrepresented, meaning the sample may not perfectly mirror the entire student population of the University of Education.
* **Pre-Filtration Impact:** Because the authors removed incomplete responses prior to publication, the dataset may contain fewer missing-value issues than would typically be expected from raw survey data.
* **Data Anonymization & Self-Reporting Bias:** The accompanying paper mentions that income-related information was collected confidentially. However, the published dataset only contains a binary indicator (`Poor_Stu`) describing whether a respondent belongs to a poor household. Additionally, many variables in the dataset are based on participant responses regarding study habits, university support, and learning experiences. Findings involving these variables should therefore be interpreted within the context of survey-based responses rather than independently verified measurements.

## 5. Data Modeling

To answer our research question, we will utilize **K-Means Clustering**. This unsupervised learning technique is the most mathematically appropriate for our objective because it evaluates our multi-dimensional feature space holistically, grouping students into cohesive, mutually exclusive profiles based on distance-based similarities. Unlike Association Rule Mining, which generates fragmented "if-then" patterns, K-Means excels at segmenting an existing population into distinct, actionable personas. This allows us to cleanly analyze the complex interactions between a student's habits and their environment.

### Preprocessing Strategy
Before applying the K-Means algorithm, the data must be properly transformed to ensure valid distance calculations. Afterward, we apply dimensionality reduction to interpret the results:

1.  **Feature Selection:** We select 11 specific features defining the internal and external student environments. The following table states what features remained and which ones were dropped. 

| Feature Name | Selected | Justification |
| :--- | :--- | :--- |
| `Study_Methods` <br> `Time_Friends` <br> `Time_SocicalMedia` <br> `Time_Studying` <br> `Adapt_Learning_Uni`| Yes | These variables define the domain of internal student behavior, adaptation, and time management, forming the first core half of our research question. |
| `Policy_Stu` <br> `SupportOf_Uni` <br> `SupportOf_Lec` <br> `Facilitie_Uni` <br> `Quality_Lecturer` <br> `TrainingCurriculum` | Yes | These variables define the domain of external institutional support and educational environment, directly addressing the second core half of our research question. |
| `GPA` | No | The target variable for academic achievement; explicitly excluded to prevent data leakage during unsupervised clustering. |
| `Competitive_Class` <br> `InfuenceF_Friends` | No | Explicitly excluded to maintain strict thematic focus on personal habits and institutional support. |
| `Year` | No | Excluded. As a static demographic variable, it falls outside the behavioral scope of our research. Additionally, the heavy skew (73.5% graduated) introduces sampling bias rather than actionable behavioral patterns. |
| `Gender` | No | Excluded due to domain irrelevance and extreme sample skew. With 88.9% of the sample being female, this static demographic variable lacks the mathematical variance required to form distinct, meaningful clusters. |
| `Minority_Stu` | No | Excluded. This is a static demographic identifier rather than an actionable behavioral or institutional metric, making it irrelevant to our clustering objectives. |
| `Poor_Stu` | No | Excluded because it is a socioeconomic factor beyond the scope of our research question, on top of this, given the dataset's data collection method, it is a self-reported metric with no clear threshold for who can be considered poor or not poor. Given the unreliability, we choose to exclude it. |
| `Father_Edu` <br> `Mother_Edu` | No | Excluded. While these static socioeconomic indicators may correlate mathematically with outcomes, they cannot be addressed by university interventions and therefore fall completely outside the actionable scope of this study. |
| `Father_Occupation` <br> `Mother_Occupation` | No | Excluded due to both domain irrelevance and data structure. As nominal categorical variables lacking inherent order, they are incompatible with distance-based clustering without heavily diluting the feature space via complex encoding. |

2.  **Binary Encoding (`Policy_Stu`):** Ten of our features are uniform ordinal variables measured on a 1-to-5 Likert scale. However, `Policy_Stu` is a nominal binary variable originally coded as 1 (Yes) and 2 (No). While scaling handles the raw math, leaving this as 1 and 2 makes post-clustering interpretation highly unintuitive. We applied **binary dummy encoding** to map this to `1 = Yes` and `0 = No`, allowing the final cluster means to be read cleanly as percentages. 
3.  **Standardization:** Since K-Means relies on Euclidean distance, we apply `StandardScaler` to ensure that each feature has a mean of 0 and a standard deviation of 1, preventing any single feature from disproportionately dominating the distance calculations.
4.  **Optimal *k* Selection:** We utilize the Elbow Method to plot the Within-Cluster Sum of Squares (WCSS) and determine the mathematically optimal number of clusters.
5.  **Dimensionality Reduction for Visualization:** Because our final model relies on 11 distinct dimensions, human visual verification is impossible on a standard graph. As a post-modeling step, we apply **Principal Component Analysis (PCA)** to mathematically compress these 11 features into a 2D space, allowing us to visually validate the separation of our clusters.

```python
from src.cbdatsi.modeling import (
    plot_elbow_curve, 
    run_kmeans_clustering, 
    plot_cluster_pca, 
    plot_cluster_profiles,
    evaluate_clusters
)

# 1. Plot Elbow Curve to determine the optimal number of clusters
plot_elbow_curve(final_df)
```

### Elbow Method Findings

The Elbow Method helps determine the optimal number of clusters by identifying the "elbow" in the WCSS curve, or the exact point where the variance stops dropping sharply. From the figure above, we can clearly observe that this elbow occurs at **k=3**. Despite seemingly multiple points that could prove to be viable given that the slope does not violently decline like typical examples shown in class, we opted to keep k=3 because it is the distinct inflection point with a gradual decline in slope afterwards.

With this parameter established, we execute the K-Means algorithm, plot the clusters utilizing PCA, and determine how well the model performed via the Silhouette Score.

```python
# 2. Execute K-Means clustering with k=3
final_df_clustered, cluster_summary = run_kmeans_clustering(final_df, n_clusters=3)

# 3. Generate PCA Visualization
plot_cluster_pca(final_df_clustered)

# 4. Evaluate the mathematical quality of the clusters
silhouette_avg = evaluate_clusters(final_df_clustered)
```

### Model Evaluation: Silhouette Score & PCA

Our model achieved a positive score of **0.1456**. A score near 0 typically indicates overlapping clusters, which perfectly aligns with what we see in our 2D PCA visualization. The clusters follow a clear, orderly progression along Principal Component 1, but they touch at their boundaries rather than forming isolated islands. 

This is exactly what we expect from human behavioral data. Because our features are based on 1-to-5 Likert scale survey responses, student habits exist on a natural, continuous spectrum rather than in perfectly separated spheres. The positive score mathematically confirms that despite this natural overlap, the algorithm successfully found valid structure: the students are more similar to their own assigned group than to their neighboring groups.

## 6. Statistical Inference

To statistically validate our research question, we must evaluate the relationship between our generated student profiles (Clusters 0, 1, and 2) and their academic achievement (`GPA`).

Because both variables are categorical, the mathematically appropriate method is the **Chi-Square Test of Independence**. This test determines whether the categorical distribution of GPA outcomes is dependent on a student's assigned cluster profile.

### Hypothesis Statement
We establish a standard significance level of $\alpha = 0.05$.

*   **Null Hypothesis ($H_0$):** GPA outcomes are independent of the student cluster profiles.
*   **Alternative Hypothesis ($H_A$):** GPA outcomes are dependent on the student cluster profiles.

### Data Preprocessing for Inference
To perform the Chi-Square Test, the raw dataset must first be preprocessed into an aggregated format. We utilize Pandas cross-tabulation (`pd.crosstab`) to transform our row-by-row student data into a **Contingency Table**, which counts the exact frequency of each GPA outcome occurring within each specific cluster.

*Note on Normality:* Normality tests apply to continuous means. Because our variables (`Cluster` and `GPA`) are categorical and evaluate frequencies, checking for a normal distribution is mathematically inapplicable and not required.

### Checking Statistical Assumptions
Before computing the p-value, we must ensure our preprocessed contingency table meets the mathematical requirements for categorical inference:
1.  **Independence:** Each observation in the table must be independent of all other observations. We already asserted this during Phase 1: each row represents a unique student who belongs to one mutually exclusive cluster and has a single GPA score.
2.  **Expected Counts:** The Chi-Square distribution appropriately estimates the sampling distribution only if every expected count in our contingency table is at least 5. We will mathematically verify this prior to interpreting the test results.

```python
from src.cbdatsi.inference import perform_chisquare_independence

# 1. Execute the Chi-Square Test of Independence with assumption check
chi2_stat, p_value, dof, obs_table = perform_chisquare_independence(
    final_df_clustered, 
    target_col='GPA', 
    cluster_col='Cluster'
)

# 2. Visualize Contigency Table as a Heatmap
print("--- Observed Contingency Table (Heatmap) ---")
display(obs_table.style.background_gradient(cmap='Blues'))
```

### Inference Conclusion

**Assumption Check Result:**
Our pre-test verification confirmed that the minimum expected count across all cross-tabulations was **19.78**, which is well above the required threshold of 5. Therefore, the assumption is met, and the Chi-Square distribution is a valid approximation for our test.

**Test of Independence Result:** 
Our Chi-Square Test of Independence yielded a $\chi^2$ statistic of **43.2357** (with 8 degrees of freedom) and a p-value of **$7.93 \times 10^{-7}$**. 

Because our resulting p-value is significantly smaller than our established significance level ($\alpha = 0.05$), we **reject the Null Hypothesis ($H_0$)**. 

**Conclusion:**
We have strong statistical evidence that a student's academic outcome is heavily dependent on their assigned profile. The differences we observed in the data are not due to random chance; the unique combinations of internal study habits and perceptions of external institutional support directly dictate the categorical distribution of a student's GPA.
