## 1. Dataset Description
**Dataset Overview & Structure**  
A group of researchers gathered demographic information, socioeconomic indicators, behavioral habits, and learning perceptions from students and alumni of the University of Education – Vietnam National University, Hanoi. The survey was designed to investigate factors that may influence student learning outcomes. Data was collected between March and June* 2023 via a Google Form questionnaire, which used the snowball sampling method. The survey originally had 2,238 responses; however, "incomplete or inappropriate" submissions were dropped prior to publication. As a result, the dataset contains **2,170 validated observations (rows)** and **22 variables (columns)**. Each row represents a survey respondent, while columns correspond to specific questions.

*Note: The accompanying paper contains a minor inconsistency regarding the data collection period. The abstract states that responses were collected from March to June 2023, while the methodology section states that the survey was conducted from March to May 2023. Since no clarification is provided by the authors, this study only notes that data collection occurred during the March–June 2023 period.

**Implications of the Data Collection Method**  
While the dataset provides a robust sample size, several limitations and design choices impact result interpretation:

* **Sampling Bias:** The survey utilized voluntary participation and snowball sampling. Highly engaged students or specific social circles may be overrepresented, meaning the sample may not perfectly mirror the entire student population of the University of Education.
* **Pre-Filtration Impact:** Because the authors removed incomplete responses prior to publication, the dataset may contain fewer missing-value issues than would typically be expected from raw survey data.
* **Data Anonymization & Self-Reporting Bias:** The accompanying paper mentions that income-related information was collected confidentially. However, the published dataset only contains a binary indicator (`Poor_Stu`) describing whether a respondent belongs to a poor household. Additionally, many variables in the dataset are based on participant responses regarding study habits, university support, and learning experiences. Findings involving these variables should therefore be interpreted within the context of survey-based responses rather than independently verified measurements.

## 2. Data Cleaning
Before exploring statistical relationships or building data models, the dataset's integrity must be structurally verified. Survey data collected via voluntary digital questionnaires often suffers from inconsistent entry formats, system-level artifacts, duplicate submissions, or structural mismatches, so we executed 4 crucial data cleaning strategies. 

1. **Structural Integrity:** Verifying that data types match the categorical scale values (coded integers).
2. **Completeness:** Screening for missing values or unrecorded answers.
3. **Uniqueness:** Identifying identical rows to distinguish system duplicates from valid, highly constrained student survey profiles.
4. **Consistency:** Confirming that categorical representations and scale limits (e.g., 1–5 rankings) remain uniform across the cohort.

* **`seaborn` (`sns`)**: An advanced statistical data plotting module built on top of Matplotlib, which enables clean, readable cross-tabulations and distributions to be rendered gracefully.

## 3. Exploratory Data Analysis (EDA)

### 3.2. Student Behavioral Time Allocation Trends
#### How do student behaviors regarding study time and social media time vary across different GPA brackets?
Since our survey responses use discrete ordinal ranks (1 to 5), standard scatter plots or boxplots look compressed and uninterpretable. We group our data by GPA brackets and normalize each bar to 100% to track the exact proportional shifts in habits. Medians and IQRs are used as robust central measures because our categorical variables are heavily skewed and non-continuous.

```python
# --- 4. BEHAVIORAL TIME ALLOCATION ---
eda1_cols = ['GPA', 'Time_Studying', 'Time_SocicalMedia']
stats1 = final_df[eda1_cols].agg(['median']).T
stats1['IQR'] = final_df[eda1_cols].quantile(0.75) - final_df[eda1_cols].quantile(0.25)

print("1. CENTRAL TENDENCY AND SPREAD SUMMARY")
display(stats1.style.format(precision=2))

print("\n2. SPEARMAN RANK CORRELATION MATRIX")
display(final_df[eda1_cols].corr(method='spearman').style.background_gradient(cmap='coolwarm', vmin=-1, vmax=1).format(precision=3))

plot_behavioral_boxplots(final_df)
```

### Summary of Observed Trends
* **Study Time:** The boxplots reveal that while a high study effort (median of 5) is claimed across all GPA brackets, top-performing students (Good and Excellent) show a much tighter, more consistent distribution exclusively clustered at the highest levels. Students in lower GPA brackets (Poor to Fair) exhibit wider variance with their distributions stretching down to lower levels, indicating less consistent study habits.

* **Social Media:** The trend clearly indicates that excessive social media use correlates with lower grades. Students in the "Poor" GPA bracket have a higher median social media usage (level 4), with the distribution heavily skewed towards the maximum. Conversely, students in the higher brackets (Good and Excellent) maintain a tighter, lower distribution (mostly concentrated between levels 2 and 3).

### 3.3. Socioeconomic Impact on Academic Performance
#### How do household wealth limitations and institutional policy support relate to a student's GPA bracket?
Because our dataset contains a baseline imbalance between cohorts, raw count bars distort comparisons. We group the data by socioeconomic tiers and normalize them horizontally to 100% to evaluate relative grade distributions. To strictly adhere to Phase 1 parameters, we analyze group-level Medians, IQRs, and calculate Spearman Rank Correlation coefficients to evaluate directional relationships among these ordinal survey categories.

```python
# --- 5. SOCIOECONOMIC IMPACT ---
for col, name in [('Poor_Stu', 'Household Wealth'), ('Policy_Stu', 'Policy Support')]:
    label_col = col + '_Label'
    
    # CHANGED: df_cleaned is now final_df
    grouped_stats = final_df.groupby(label_col)['GPA'].agg(Median='median', IQR=lambda x: x.quantile(0.75) - x.quantile(0.25))
    
    print(f"\nGPA Center & Spread by {name}:")
    display(grouped_stats.style.format(precision=2))

plot_socioeconomic_conditional(final_df)
```

### Summary of Observed Trends
* **Household Wealth:** By analyzing the breakdown within each individual GPA bracket, we can see that the proportion of students from poor households remains relatively flat and consistent (ranging roughly between 2% and 5%) across all performance levels. This suggests that financial status alone does not disproportionately drive students into the lowest academic tiers.

* **Policy Support:** The conditional distribution shows that students receiving policy support make up the largest proportions within the Average (approx. 44%) and Fair (approx. 40%) GPA brackets, rather than at the extremes. This indicates that institutional policy assistance is effectively doing its job—acting as a safety net that keeps at-risk students competitive and firmly afloat within the passing and middle tiers.

### 3.4. Institutional Influence on Academic Outcomes
#### Do institutional support frameworks and structural quality perceptions show a meaningful relationship with student GPA tiers?
Because perceptual survey data often displays strong internal collinearity, a full correlation matrix clutters the visualization with irrelevant internal metrics. To maintain a strict focus on our target outcome, we compute independent Spearman rank correlation coefficients exclusively against GPA and isolate them into a single-column array. This is much cleaner as opposed to opting for a full heatmap.

```python
# --- 6. INSTITUTIONAL INFLUENCE ---
inst_cols = ['SupportOf_Uni', 'SupportOf_Lec', 'Facilitie_Uni', 'Quality_Lecturer', 'TrainingCurriculum', 'GPA']
inst_corr = final_df[inst_cols].corr(method='spearman')[['GPA']].drop('GPA').sort_values(by='GPA', ascending=False)

print("--- Sorted Institutional Factors Correlation to GPA ---")
display(inst_corr.style.format(precision=3))

plot_institutional_heatmap(final_df)
```

### Summary of Observed Trends
* **Quality of Lecturer:** The quality of the lecturer has the strongest link to high grades. However, the connection is still quite small, meaning great teaching is just one piece of a much bigger puzzle rather than the sole driver of success.

* **University Infrastructure:** A university's physical buildings and general administrative support have the absolute weakest connection to student grades. The campus environment matters, but it doesn't move the needle much on its own.

* **Other Institutional metrics:** The quality of the curriculum and direct help from teachers sit right in the middle. They have a small but noticeable positive impact on student performance.

While no single school factor dictates a student's GPA on its own, they add up. A university's true impact can't be pinned on just one variable—it is the collection of all these small pieces working together that builds a great learning environment.

### 3.5. Internal Mindset vs. External Pressures
#### Do students' internal study habits and adaptability impact their GPA more heavily than external peer and classroom pressures?
To evaluate these distinct behavioral domains, we contrast internal psychological factors (adaptation to university, study methods) against external environmental pressures (class competition, friend influence). We compute Spearman Rank Correlations for each variable against GPA, lastly we visualize using a Normalized Cross-Tabulation Heatmap.

```python
# --- 7. INTERNAL VS EXTERNAL PRESSURES ---
pressure_cols = ['Adapt_Learning_Uni', 'Study_Methods', 'Competitive_Class', 'InfuenceF_Friends', 'GPA']
pressure_corr = final_df[pressure_cols].corr(method='spearman')[['GPA']].drop('GPA').sort_values(by='GPA', ascending=False)

print("--- Spearman Correlation: Internal vs External Pressures ---")
display(pressure_corr.style.format(precision=3))

plot_mindset_heatmaps(final_df)
```

Personal factors like good study habits and how well you adapt to college life have a clear, direct link to higher grades. As these skills improve, a student's GPA visibly climbs.
* **External Pressures:** Conversely, external factors like *Class Competition* (*0.032*) and *Friend Influence* (*0.001*) show a highly scattered, flat distribution across the board. External factors like competition in class or influence from friends have almost zero impact on grades. High social pressure or a competitive environment doesn't reliably change a student's GPA tier.

The data clearly indicates that what happens inside a student matters vastly more than what happens around them. Personal academic habits and resilience are far better predictors of success than social or competitive environments.
