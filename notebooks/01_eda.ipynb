## 1. Dataset Description
**Dataset Overview & Structure**  
A group of researchers gathered demographic information, socioeconomic indicators, behavioral habits, and learning perceptions from students and alumni of the University of Education – Vietnam National University, Hanoi. The survey was designed to investigate factors that may influence student learning outcomes. Data was collected between March and June* 2023 via a Google Form questionnaire, which used the snowball sampling method. The survey originally had 2,238 responses; however, "incomplete or inappropriate" submissions were dropped prior to publication. As a result, the dataset contains **2,170 validated observations (rows)** and **22 variables (columns)**. Each row represents a survey respondent, while columns correspond to specific questions.

*Note: The accompanying paper contains a minor inconsistency regarding the data collection period. The abstract states that responses were collected from March to June 2023, while the methodology section states that the survey was conducted from March to May 2023. Since no clarification is provided by the authors, this study only notes that data collection occurred during the March–June 2023 period.

**Implications of the Data Collection Method**  
While the dataset provides a robust sample size, several limitations and design choices impact result interpretation:

* **Sampling Bias:** The survey utilized voluntary participation and snowball sampling. Highly engaged students or specific social circles may be overrepresented, meaning the sample may not perfectly mirror the entire student population of the University of Education.
* **Pre-Filtration Impact:** Because the authors removed incomplete responses prior to publication, the dataset may contain fewer missing-value issues than would typically be expected from raw survey data.
* **Data Anonymization & Self-Reporting Bias:** The accompanying paper mentions that income-related information was collected confidentially. However, the published dataset only contains a binary indicator (`Poor_Stu`) describing whether a respondent belongs to a poor household. Additionally, many variables in the dataset are based on participant responses regarding study habits, university support, and learning experiences. Findings involving these variables should therefore be interpreted within the context of survey-based responses rather than independently verified measurements.

Survey data collected via voluntary digital questionnaires often suffers from inconsistent entry formats, system-level artifacts, duplicate submissions, or structural mismatches, so we did 4 crucial data cleaning strategies. 

1. **Structural Integrity:** Verifying that data types match the categorical scale values (coded integers).
2. **Completeness:** Screening for missing values or unrecorded answers.

### 1. Demographic and Target Baseline Overview
#### What is the structural composition of our student sample across year level, gender, and baseline GPA performance?
#### Before examining behavioral relationships, we explore the univariate distribution of our sample to understand its baseline characteristics across academic year levels, gender distribution, and our target variable (`GPA`).

```python
# --- 1. DATA PREPARATION ---
year_map = {1: 'First-year', 2: 'Second-year', 3: 'Third-year', 4: 'Fourth-year', 5: 'Graduated'}
gpa_map = {1: 'Poor', 2: 'Average', 3: 'Fair', 4: 'Good', 5: 'Excellent'}

df['Year_Label'] = df['Year'].map(year_map)
df['Gender_Label'] = df['Gender'].map({1: 'Male', 2: 'Female'})
df['GPA_Label'] = df['GPA'].map(gpa_map)

# --- 2. BASELINE SUMMARY TABLES ---
print("--- Baseline Respondent Distributions ---")
for col, name in [('Year_Label', 'Year Level'), ('Gender_Label', 'Gender'), ('GPA_Label', 'GPA Bracket')]:
    counts = df[col].value_counts()
    percs = df[col].value_counts(normalize=True).mul(100).round(1).astype(str) + '%'
    display(pd.DataFrame({'Count': counts, 'Percentage': percs}).style.set_caption(f"Distribution of {name}"))

# --- 3. PURE MATPLOTLIB VISUALIZATION ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Helper to plot native bar distributions in an ordered sequence
def plot_univariate_bar(ax, series, title, order=None, color='#2F5597'):
    counts = series.value_counts()
    if order:
        counts = counts.reindex(order).fillna(0)
    ax.bar(counts.index.astype(str), counts.values, color=color, edgecolor='black', alpha=0.85)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_ylabel('Number of Students')
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.tick_params(axis='x', rotation=15)

# Plot A: Year Level
year_order = ['First-year', 'Second-year', 'Third-year', 'Fourth-year', 'Graduated']
plot_univariate_bar(axes[0], df['Year_Label'], 'Distribution by Year Level', order=year_order, color='#2F5597')

# Plot B: Gender
plot_univariate_bar(axes[1], df['Gender_Label'], 'Distribution by Gender', color='#8FAADC')

# Plot C: Target Variable (GPA)
gpa_order = ['Poor', 'Average', 'Fair', 'Good', 'Excellent']
plot_univariate_bar(axes[2], df['GPA_Label'], 'Baseline GPA Distribution', order=gpa_order, color='#4472C4')

plt.tight_layout()
plt.show()
```

### 3. Socioeconomic Impact on Academic Performance
#### How do household wealth limitations and institutional policy support relate to a student's GPA bracket?
#### Because our dataset contains a baseline imbalance between cohorts, raw count bars distort comparisons. We group the data by socioeconomic tiers and normalize them horizontally to 100% to evaluate relative grade distributions. To strictly adhere to Phase 1 parameters, we analyze group-level Medians, IQRs, and calculate Spearman Rank Correlation coefficients to evaluate directional relationships among these ordinal survey categories.

```python
# --- 1. DATA PREPARATION & REGENERATION ---
gpa_map = {1: 'Poor', 2: 'Average', 3: 'Fair', 4: 'Good', 5: 'Excellent'}
df['GPA_Label'] = df['GPA'].map(gpa_map)
df['Poor_Stu_Label'] = df['Poor_Stu'].map({1: 'Yes (Poor)', 2: 'No (Not Poor)'})
df['Policy_Stu_Label'] = df['Policy_Stu'].map({1: 'Yes (Supported)', 2: 'No (Not Supported)'})

# --- 2. DESCRIPTIVE STATS---
print("--- Phase 1 Descriptive Measures by Socioeconomic Factors ---")
for col, name in [('Poor_Stu', 'Household Wealth'), ('Policy_Stu', 'Policy Support')]:
    # Grouped Central Tendency (Median) and Dispersion (IQR)
    label_col = col + '_Label'
    grouped_stats = df.groupby(label_col)['GPA'].agg(
        Median='median',
        IQR=lambda x: x.quantile(0.75) - x.quantile(0.25)
    )
    
    # Spearman Correlation
    spearman_corr = df[col].corr(df['GPA'], method='spearman')
    
    print(f"\n{name} Analysis:")
    display(grouped_stats.style.format(precision=2).set_caption(f"GPA Center & Spread by {name}"))
    print(f"Spearman Rank Correlation with GPA: {spearman_corr:.3f}")

# --- 3. NORMALIZED PROPORTIONAL VISUALIZATION ---
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

gpa_order = ['Poor', 'Average', 'Fair', 'Good', 'Excellent']
dark_gpa_colors = ['#8C2D19', '#B37D14', '#4A5568', '#1F4E79', '#134E3A']

# Plot A: Normalized Household Wealth Breakdown
ct_wealth = pd.crosstab(df['Poor_Stu_Label'], df['GPA_Label'], normalize='index') * 100
ct_wealth = ct_wealth.reindex(columns=gpa_order)
ct_wealth.plot(kind='bar', ax=axes[0], color=dark_gpa_colors, edgecolor='black', width=0.8)

axes[0].set_title('Proportional GPA Breakdown by Household Wealth', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Financial Status', fontsize=11)
axes[0].set_ylabel('Percentage of Cohort (%)', fontsize=11)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].legend(title='GPA Bracket', loc='upper right')
axes[0].yaxis.set_major_formatter('{x:.0f}%')
axes[0].set_ylim(0, 75)

for p in axes[0].patches:
    height = p.get_height()
    if height > 0:
        axes[0].annotate(f'{height:.1f}%', 
                         (p.get_x() + p.get_width() / 2., height), 
                         ha='center', va='bottom', 
                         xytext=(0, 3), 
                         textcoords='offset points', fontsize=8, fontweight='bold')

# Plot B: Normalized Policy Support Breakdown
ct_policy = pd.crosstab(df['Policy_Stu_Label'], df['GPA_Label'], normalize='index') * 100
ct_policy = ct_policy.reindex(columns=gpa_order)
ct_policy.plot(kind='bar', ax=axes[1], color=dark_gpa_colors, edgecolor='black', width=0.8)

axes[1].set_title('Proportional GPA Breakdown by Policy Support', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Policy Assistance', fontsize=11)
axes[1].set_ylabel('Percentage of Cohort (%)', fontsize=11)
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=0)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].legend(title='GPA Bracket', loc='upper right')
axes[1].yaxis.set_major_formatter('{x:.0f}%')
axes[1].set_ylim(0, 75)

for p in axes[1].patches:
    height = p.get_height()
    if height > 0:
        axes[1].annotate(f'{height:.1f}%', 
                         (p.get_x() + p.get_width() / 2., height), 
                         ha='center', va='bottom', 
                         xytext=(0, 3), 
                         textcoords='offset points', fontsize=8, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Summary of Observed Trends
* **Household Wealth:** Proportions of **Excellent** grades drop significantly from **5%** in the **Non-poor** group down to just **2.3%** for the **Poor** cohort, showing a clear economic barrier at the absolute top tier. This phenomenon extends to **Good** performing students but interestingly, students categorized as **Poor** maintain a much better baseline performance at **Fair** with far less students in the **Average** and **Poor** GPA brackets compared to the **Non-poor** students.
* **Policy Support:** Policy assistance displays an unexpected pattern where unsupported students actually maintain a much larger share of **Excellent** marks (**5.4%**) compared to supported students (**4%**). This suggests that institutional aid might be effective at targetting students who are facing greater academic risk or adversity. However, the consistent stability in the **Fair** tier (**61.6%** vs **51.1%**) shows that policy-supported students remain highly competitive and heavily clustered around strong passing averages. Another great thing to note is the gap between supported and non-supported students is smaller, implying that those financially burdened students of whom the policy targets do experience positive academic benefits, closing the gap between the less privileged students and those who are financially sound.

### 4. Institutional Influence on Academic Outcomes
#### Do institutional support frameworks and structural quality perceptions show a meaningful relationship with student GPA tiers?
#### Because perceptual survey data often displays strong internal collinearity, a full correlation matrix clutters the visualization with irrelevant internal metrics. To maintain a strict focus on our target outcome, we compute independent Spearman rank correlation coefficients exclusively against GPA and isolate them into a single-column array. This is much cleaner as opposed to opting for a full heatmap.

```python
# --- 1. DATA PREPARATION & BASELINE EXTRACTION ---
inst_cols = ['SupportOf_Uni', 'SupportOf_Lec', 'Facilitie_Uni', 'Quality_Lecturer', 'TrainingCurriculum', 'GPA']

# Compute the global Spearman rank correlation matrix
inst_corr = df[inst_cols].corr(method='spearman')

# Isolate ONLY the relationship with GPA and sort from strongest to weakest connection
gpa_corr_focused = inst_corr[['GPA']].drop('GPA').sort_values(by='GPA', ascending=False)

# --- 2. TABLE OF VALUES ---
print("--- Sorted Spearman Rank Correlation Coefficients with Student GPA ---")
display(gpa_corr_focused.style.format(precision=3).set_caption("Institutional Factors Correlation to GPA"))

# --- 3. TARGET-FOCUSED HEATMAP ---
plt.figure(figsize=(6, 5))

sns.heatmap(
    gpa_corr_focused, 
    annot=True, 
    cmap='coolwarm', 
    fmt=".3f", 
    vmin=-0.2, vmax=0.2, 
    linewidths=1.5, 
    linecolor='black',
    cbar_kws={"label": "Spearman Correlation Coefficient (ρ)", "shrink": 0.9}
)

plt.title('Target Correlation Matrix:\nInstitutional Factors vs. Student GPA', fontsize=12, fontweight='bold', pad=15)
plt.ylabel('Institutional Perception Metric', fontsize=11)
plt.xlabel('Target Variable', fontsize=11)
plt.xticks([0.5], ['GPA'])

plt.tight_layout()
plt.show()
```

### Summary of Observed Trends
* **Quality of Lecturer:** The **Quality_Lecturer** factor yields the highest correlation value at **0.102**, showing a slight positive trend indicating that higher perceived quality of instructional execution tracks alongside upper tier grades. Though we need to be cautious as an effect size near **0.10** indicates a very weak relationship, implying that while teaching quality matters, it operates as part of a much larger web of academic influences rather than acting as a singular driver.
* **University Infrastructure:** The structural aspects such as **SupportOf_Uni** (**0.053**) and **Facilitie_Uni** (**0.048**) display the absolute weakest correlations to student outcomes within this domain. This signals that physical institutional environments and macro-level administrative systems possess a weak but possibly relevant impact to the academic performance of students.
* **Other Institutional metrics:** Not too far above, though a decent amount more impactful than the weakest correlations are TrainingCurriculum (**0.057**) and SupportOf_Lec(**0.066**), this implies that the direct involvement of lecturers and the quality of the curriculum has a weak but notable impact on the individual academic performance of students.

The important thing to note here is that while individual variables may show weak, almost insignificant correlation to GPA, the combination of these factors may yield and overall significant and noticeable impact to the academic performance to students, meaning an institution's impact cannot be simply condensed into one variable, rather it is a collection of little things that build up an institution's excellence.

### 5. Internal Mindset vs. External Pressures
#### Do students' internal study habits and adaptability impact their GPA more heavily than external peer and classroom pressures?
#### To evaluate these distinct behavioral domains, we contrast internal psychological factors (adaptation to university, study methods) against external environmental pressures (class competition, friend influence). We compute Spearman Rank Correlations for each variable against GPA, lastly we visualize using a Normalized Cross-Tabulation Heatmap.

```python
# --- 1. DESCRIPTIVE STATS: SPEARMAN CORRELATION ---
pressure_cols = ['Adapt_Learning_Uni', 'Study_Methods', 'Competitive_Class', 'InfuenceF_Friends', 'GPA']

print("--- Spearman Correlation: Internal vs External Pressures ---")
pressure_corr = df[pressure_cols].corr(method='spearman')
gpa_pressure = pressure_corr[['GPA']].drop('GPA').sort_values(by='GPA', ascending=False)
display(gpa_pressure.style.format(precision=3).set_caption("Correlation to GPA"))

# --- 2. VISUALIZATION: NORMALIZED CROSSTAB HEATMAPS ---
# We use crosstabs normalized by columns to see the GPA breakdown FOR EACH survey response level
ct_study = pd.crosstab(df['GPA'], df['Study_Methods'], normalize='columns')
ct_comp = pd.crosstab(df['GPA'], df['Competitive_Class'], normalize='columns')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot A: Internal Mindset (Study Methods)
sns.heatmap(ct_study, annot=True, fmt=".0%", cmap="Blues", 
            cbar_kws={'label': '% of Students in GPA Bracket'}, ax=axes[0])
axes[0].set_title('GPA Distribution by Study Methods (Internal)', fontsize=12, fontweight='bold', pad=15)
axes[0].set_xlabel('Study Methods Score (1 = Lowest, 5 = Highest)')
axes[0].set_ylabel('GPA Bracket')
axes[0].invert_yaxis() # Puts Excellent (5) at the top

# Plot B: External Pressures (Class Competition)
sns.heatmap(ct_comp, annot=True, fmt=".0%", cmap="Oranges", 
            cbar_kws={'label': '% of Students in GPA Bracket'}, ax=axes[1])
axes[1].set_title('GPA Distribution by Class Competition (External)', fontsize=12, fontweight='bold', pad=15)
axes[1].set_xlabel('Class Competition Score (1 = Lowest, 5 = Highest)')
axes[1].set_ylabel('GPA Bracket')
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()
```

### Summary of Observed Trends
* **Internal Mindset:** The normalized distribution reveals a clear directional shift for internal factors like **Study Methods** (**0.155**) and **Adaptation to Uni** (**0.146**). As students report higher internal scores, the concentration of students visibly shifts upward into the higher GPA brackets. This indicates a positive structural relationship between strong personal academic routines and final grades.
* **External Pressures:** Conversely, external factors like **Class Competition** (**0.032**) and **Friend Influence** (**0.001**) show a highly scattered, flat distribution across the board. High external pressure scores do not yield a cohesive concentration of upper-tier GPAs, which perfectly aligns with their near-zero correlation coefficients. 

The data clearly indicates that a student's internal academic resilience and personal study habits are much more reliable indicators of their GPA tier than the competitive or social nature of their immediate external environment.
