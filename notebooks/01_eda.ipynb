## 1. Dataset Description
**Dataset Overview & Structure**  
A group of researchers gathered demographic information, socioeconomic indicators, behavioral habits, and learning perceptions from students and alumni of the University of Education – Vietnam National University, Hanoi. The survey was designed to investigate factors that may influence student learning outcomes. Data was collected between March and June* 2023 via a Google Form questionnaire, which used the snowball sampling method. The survey originally had 2,238 responses; however, "incomplete or inappropriate" submissions were dropped prior to publication. As a result, the dataset contains **2,170 validated observations (rows)** and **22 variables (columns)**. Each row represents a survey respondent, while columns correspond to specific questions.

*Note: The accompanying paper contains a minor inconsistency regarding the data collection period. The abstract states that responses were collected from March to June 2023, while the methodology section states that the survey was conducted from March to May 2023. Since no clarification is provided by the authors, this study only notes that data collection occurred during the March–June 2023 period.

**Implications of the Data Collection Method**  
While the dataset provides a robust sample size, several limitations and design choices impact result interpretation:

* **Sampling Bias:** The survey utilized voluntary participation and snowball sampling. Highly engaged students or specific social circles may be overrepresented, meaning the sample may not perfectly mirror the entire student population of the University of Education.
* **Pre-Filtration Impact:** Because the authors removed incomplete responses prior to publication, the dataset may contain fewer missing-value issues than would typically be expected from raw survey data.
* **Data Anonymization & Self-Reporting Bias:** The accompanying paper mentions that income-related information was collected confidentially. However, the published dataset only contains a binary indicator (`Poor_Stu`) describing whether a respondent belongs to a poor household. Additionally, many variables in the dataset are based on participant responses regarding study habits, university support, and learning experiences. Findings involving these variables should therefore be interpreted within the context of survey-based responses rather than independently verified measurements.

```python
# --- EDA QUESTION 1: TIME ALLOCATION ---

# 1. Descriptive Stats
eda1_cols = ['Time_Studying', 'Time_SocicalMedia', 'GPA']
print("--- Median and IQR for Time Allocation ---")

# Calculate Median and IQR
stats1 = df[eda1_cols].agg(['median']).T
stats1['IQR'] = df[eda1_cols].apply(lambda x: x.quantile(0.75) - x.quantile(0.25))
display(stats1)

print("\n--- Spearman Rank Correlation ---")
# numeric_only=True prevents any string/float crashes!
display(df[eda1_cols].corr(method='spearman', numeric_only=True).round(3))

# 2. Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot A: Study Time 
# (medianprops forces a thick, bright orange median line)
sns.boxplot(data=df, x='Time_Studying', y='GPA', ax=axes[0], palette='Blues', 
            medianprops={'color': 'orange', 'linewidth': 2.5})
axes[0].set_title('GPA Distribution by Study Time')
axes[0].set_xlabel('Study Time (1 = Low, 5 = High)')
axes[0].set_ylabel('GPA Bracket (1=Poor, 5=Excellent)')

# Plot B: Social Media Time
sns.boxplot(data=df, x='Time_SocicalMedia', y='GPA', ax=axes[1], palette='Reds', 
            medianprops={'color': 'orange', 'linewidth': 2.5})
axes[1].set_title('GPA Distribution by Social Media Time')
axes[1].set_xlabel('Social Media Time (1 = Low, 5 = High)')
axes[1].set_ylabel('GPA Bracket (1=Poor, 5=Excellent)')

plt.tight_layout()
plt.show()
```

```python
# --- EDA QUESTION 2: SOCIOECONOMIC BASELINES ---

# 1. Regenerate Labels (Protects against Jupyter out-of-order execution)
gpa_map = {1: 'Poor', 2: 'Average', 3: 'Fair', 4: 'Good', 5: 'Excellent'}
df['GPA_Label'] = df['GPA'].map(gpa_map)

poor_map = {1: 'Yes (Poor)', 2: 'No (Not Poor)'}
df['Poor_Stu_Label'] = df['Poor_Stu'].map(poor_map)

policy_map = {1: 'Yes (Supported)', 2: 'No (Not Supported)'}
df['Policy_Stu_Label'] = df['Policy_Stu'].map(policy_map)

# 2. Descriptive Stats: Grouped Central Tendency
print("--- Median GPA & IQR by Socioeconomic Factors ---")
for col in ['Poor_Stu', 'Policy_Stu']:
    grouped_stats = df.groupby(col)['GPA'].agg(
        Median='median',
        IQR=lambda x: x.quantile(0.75) - x.quantile(0.25)
    ).round(2)
    print(f"\nStats for {col} (1=Yes, 2=No):")
    display(grouped_stats)

# 3. Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot A: Household Wealth
sns.countplot(data=df, x='GPA_Label', hue='Poor_Stu_Label', 
              order=['Poor', 'Average', 'Fair', 'Good', 'Excellent'], ax=axes[0], palette='Set2')
axes[0].set_title('GPA Distribution by Household Wealth')
axes[0].set_xlabel('GPA Bracket')
axes[0].set_ylabel('Number of Students')

# Plot B: Policy Support
sns.countplot(data=df, x='GPA_Label', hue='Policy_Stu_Label', 
              order=['Poor', 'Average', 'Fair', 'Good', 'Excellent'], ax=axes[1], palette='Set2')
axes[1].set_title('GPA Distribution by Policy Support')
axes[1].set_xlabel('GPA Bracket')
axes[1].set_ylabel('Number of Students')

plt.tight_layout()
plt.show()
```

```python
# --- EDA QUESTION 3: INSTITUTIONAL EFFECTIVENESS ---

# 1. Descriptive Stats: Institutional Correlation
inst_cols = ['SupportOf_Uni', 'SupportOf_Lec', 'Facilitie_Uni', 'Quality_Lecturer', 'TrainingCurriculum', 'GPA']
print("--- Spearman Correlation: Institutional Factors vs GPA ---")

# Calculate correlation strictly on these ordinal columns to prevent crashes
inst_corr = df[inst_cols].corr(method='spearman')
display(inst_corr['GPA'].drop('GPA').sort_values(ascending=False).round(3))

# 2. Visualization: Focused Heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(inst_corr, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1, 
            linewidths=0.5, cbar_kws={"shrink": .8})
plt.title('Correlation Heatmap: Institutional Perception vs. GPA', pad=15)
plt.tight_layout()
plt.show()

# INSTRUCTIONS FOR WRITING YOUR INSIGHT:
# Look at the bottom row (or right column) of the heatmap where variables intersect with GPA. 
# Which box is the darkest red (highest number)? Write your finding below this block.
```

```python
# --- EDA QUESTION 4: INTERNAL VS EXTERNAL PRESSURES ---

# 1. Descriptive Stats
pressure_cols = ['Adapt_Learning_Uni', 'Study_Methods', 'Competitive_Class', 'InfuenceF_Friends', 'GPA']
print("--- Spearman Correlation: Internal vs External Pressures ---")

# Calculate correlation and sort it
pressure_corr = df[pressure_cols].corr(method='spearman')
display(pressure_corr['GPA'].drop('GPA').sort_values(ascending=False).round(3))

# 2. Visualization: Pointplots to show trajectories
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot A: Internal Factors
sns.pointplot(data=df, x='Study_Methods', y='GPA', color='mediumblue', label='Study Methods', ax=axes[0])
sns.pointplot(data=df, x='Adapt_Learning_Uni', y='GPA', color='deepskyblue', label='Adaptation to Uni', ax=axes[0])
axes[0].set_title('GPA Trajectory by Internal Mindset')
axes[0].set_xlabel('Score (1 = Lowest, 5 = Highest)')
axes[0].set_ylabel('Average GPA')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].legend()

# Plot B: External Pressures
sns.pointplot(data=df, x='Competitive_Class', y='GPA', color='firebrick', label='Class Competition', ax=axes[1])
sns.pointplot(data=df, x='InfuenceF_Friends', y='GPA', color='darkorange', label='Friend Influence', ax=axes[1])
axes[1].set_title('GPA Trajectory by External Pressures')
axes[1].set_xlabel('Score (1 = Lowest, 5 = Highest)')
axes[1].set_ylabel('Average GPA')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].legend()

plt.tight_layout()
plt.show()

# INSTRUCTIONS FOR WRITING YOUR INSIGHT:
# Look at the point plots. A steeper upward line means a stronger impact. 
# Do internal factors (blue lines) have a steeper, more consistent climb than external pressures (red/orange lines)?
```
